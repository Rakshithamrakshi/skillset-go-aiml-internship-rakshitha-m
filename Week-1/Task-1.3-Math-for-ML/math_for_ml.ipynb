# Task 1.3 - Math for ML: Worked Solutions

This notebook covers descriptive statistics, probability, linear algebra, 
and basic calculus concepts, connected to their applications in machine learning.

```python
import numpy as np

# --- Descriptive Statistics ---
scores = np.array([85, 90, 78, 92, 88, 76, 95, 89, 84, 91])

mean = np.mean(scores)
median = np.median(scores)
variance = np.var(scores)
std_dev = np.std(scores)

print("Data:", scores)
print(f"Mean: {mean}")
print(f"Median: {median}")
print(f"Variance: {variance:.2f}")
print(f"Standard Deviation: {std_dev:.2f}")
```

**ML Connection:** Mean and standard deviation are the foundation of **feature 
standardization** (`z = (x - mean) / std_dev`), a preprocessing step used in 
almost every ML pipeline to bring features onto a comparable scale before 
training a model. Without this, features with larger numeric ranges can 
dominate distance-based or gradient-based algorithms.

```python
# --- Covariance and Correlation ---
study_hours = np.array([2, 4, 5, 3, 6, 7, 1, 8])
exam_scores = np.array([55, 65, 70, 60, 75, 85, 50, 90])

covariance_matrix = np.cov(study_hours, exam_scores)
correlation_matrix = np.corrcoef(study_hours, exam_scores)

print("Covariance matrix:\n", covariance_matrix)
print("\nCovariance between study_hours and exam_scores:", covariance_matrix[0, 1])
print("\nCorrelation matrix:\n", correlation_matrix)
print("\nCorrelation coefficient:", correlation_matrix[0, 1])
```

**Important Caution:** A correlation of 0.99 does NOT mean study hours *cause* 
higher scores — it only means they move together in this data. Correlation 
never proves causation; a hidden factor (e.g. general discipline) could 
influence both. In ML, this matters because a model can find strong 
correlations that are coincidental or driven by confounding variables, not 
genuine cause-effect relationships.

**ML Connection:** Correlation is used in feature selection — highly 
correlated features often carry redundant information, and checking a 
correlation matrix helps decide which features to keep or drop before 
training a model.

```python
# --- Probability ---
import random

# Simple example: rolling a die, checking conditional probability
outcomes = [1, 2, 3, 4, 5, 6]

# P(even number)
even_numbers = [x for x in outcomes if x % 2 == 0]
p_even = len(even_numbers) / len(outcomes)
print(f"P(even) = {p_even}")

# P(greater than 4)
greater_than_4 = [x for x in outcomes if x > 4]
p_greater_than_4 = len(greater_than_4) / len(outcomes)
print(f"P(greater than 4) = {p_greater_than_4}")

# Independence check: P(even AND greater than 4)
both = [x for x in outcomes if x % 2 == 0 and x > 4]
p_both = len(both) / len(outcomes)
print(f"P(even AND greater than 4) = {p_both}")
print(f"P(even) * P(greater than 4) = {p_even * p_greater_than_4:.3f}")
print("If these two values are equal, the events are independent.")

# Expected value example: a simple weighted average
values = [10, 20, 30]
probabilities = [0.5, 0.3, 0.2]
expected_value = sum(v * p for v, p in zip(values, probabilities))
print(f"\nExpected value = {expected_value}")
```

**ML Connection:** Expected value underlies loss functions — training loss is 
essentially the *expected* error across the dataset. Probability distributions 
also appear directly in classification models, where outputs like softmax 
produce a probability for each class.

```python
# --- Vectors, Matrices, Dot Product ---
import numpy as np
vector_a = np.array([2, 3, 1])
vector_b = np.array([4, 0, 5])

dot_product = np.dot(vector_a, vector_b)
print(f"Vector A: {vector_a}, shape: {vector_a.shape}")
print(f"Vector B: {vector_b}, shape: {vector_b.shape}")
print(f"Dot product: {dot_product}")

# Matrix example
matrix_X = np.array([[1, 2], [3, 4]])
matrix_W = np.array([[5, 6], [7, 8]])

matrix_product = np.dot(matrix_X, matrix_W)
print(f"\nMatrix X:\n{matrix_X}, shape: {matrix_X.shape}")
print(f"Matrix W:\n{matrix_W}, shape: {matrix_W.shape}")
print(f"Matrix product (X . W):\n{matrix_product}")
```

**ML Connection:** A single neuron's output is literally a dot product: 
`output = (weights . inputs) + bias`. Every layer in a neural network is a 
matrix multiplication between inputs and weights — this is exactly why 
vector/matrix dimension compatibility is critical when designing networks.

```python
# --- Derivatives and Gradient Descent Intuition ---

def f(x):
    return x**2  # a simple function: loss-like curve

def derivative_f(x):
    return 2*x  # derivative of x^2 is 2x

# Gradient descent: start at a point, move opposite to the gradient
x = 10.0  # starting point (like a bad initial guess)
learning_rate = 0.1

print("Gradient Descent Steps:")
for step in range(10):
    grad = derivative_f(x)
    x = x - learning_rate * grad
    print(f"Step {step+1}: x = {x:.4f}, f(x) = {f(x):.4f}")

print(f"\nFinal x (should be close to 0, the minimum): {x:.4f}")
```

**ML Connection:** This tiny loop is the core of how every neural network 
learns. Real training replaces `f(x)=x²` with a loss function over millions 
of parameters, but the idea is identical: compute the gradient, step opposite 
to it, repeat.
